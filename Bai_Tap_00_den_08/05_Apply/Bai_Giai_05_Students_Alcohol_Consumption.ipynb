# Bài 05: Áp Dụng Hàm Số Trong Pandas - Apply (Học Sinh & Tiêu Thụ Cồn)
### Giới thiệu:
Thư viện Pandas cung cấp phương thức .apply() cực kỳ mạnh mẽ để áp dụng bất kỳ hàm tự tạo nào (Custom Function hoặc Lambda Function) lên từng hàng, từng cột hoặc từng phần tử của bảng dữ liệu:
- Cắt lát bảng dữ liệu theo nhãn cột bằng .loc[:, 'start':'end'].
- Viết và sử dụng hàm ẩn danh lambda.
- Hiểu tính bất biến (Immutability) khi sử dụng .apply() và cách gán đè để cập nhật dữ liệu.
- Viết hàm logic phân nhánh với if...else và tạo ra đặc trưng cột mới.
- Áp dụng hàm biến đổi trên toàn bộ bảng dữ liệu (df.map() / df.applymap()).


### Step 1. Import các thư viện cần thiết

```python
import pandas as pd
import numpy as np
```

### Step 2 & 3. Đọc dữ liệu từ URL và gán vào biến df

```python
url = 'https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/student-mat.csv'
df = pd.read_csv(url)
df.head()
```

### Step 4. Cắt lát DataFrame từ cột 'school' cho đến cột 'guardian'
*Sử dụng .loc[:, 'school':'guardian'] để lấy toàn bộ các hàng và dải cột từ school đến guardian.*

```python
df = df.loc[:, 'school':'guardian']
df.head()
```

### Step 5. Tạo một lambda function viết hoa chữ cái đầu của chuỗi (capitalize)

```python
capitalize_func = lambda x: x.capitalize() if isinstance(x, str) else x
```

### Step 6. Áp dụng hàm viết hoa cho cả hai cột Mjob (nghề của mẹ) và Fjob (nghề của bố)

```python
df['Mjob'].apply(capitalize_func)
df['Fjob'].apply(capitalize_func)
```

### Step 7. In ra các dòng cuối cùng của tập dữ liệu

```python
df.tail()
```

### Step 8. Tại sao DataFrame gốc vẫn là chữ thường? Khắc phục điều này!
**Giải thích:** Phương thức .apply() trả về một Series mới chứa kết quả biến đổi chứ KHÔNG trực tiếp thay đổi DataFrame gốc (không có cơ chế inplace).  
**Cách khắc phục:** Ta phải gán ngược kết quả trả về vào chính các cột đó: df['col'] = df['col'].apply(...).

```python
df['Mjob'] = df['Mjob'].apply(capitalize_func)
df['Fjob'] = df['Fjob'].apply(capitalize_func)
df.tail()
```

### Step 9. Tạo hàm tên là majority trả về kiểu boolean cho cột mới 'legal_drinker' (Tuổi > 17 được coi là đủ tuổi uống đồ có cồn)

```python
def majority(age):
    return age > 17

df['legal_drinker'] = df['age'].apply(majority)
df.head()
```

### Step 10. Nhân tất cả các giá trị số trong tập dữ liệu với 10 (giữ nguyên chuỗi)
*Sử dụng .map() (hoặc .applymap() trong pandas cũ) để biến đổi từng ô dữ liệu.*

```python
def multiply_10(val):
    if isinstance(val, (int, float, np.number)) and not isinstance(val, bool):
        return val * 10
    return val

# Áp dụng hàm lên từng ô của DataFrame
if hasattr(df, 'map'):
    df_multiplied = df.map(multiply_10)
else:
    df_multiplied = df.applymap(multiply_10)

df_multiplied.head()
```
