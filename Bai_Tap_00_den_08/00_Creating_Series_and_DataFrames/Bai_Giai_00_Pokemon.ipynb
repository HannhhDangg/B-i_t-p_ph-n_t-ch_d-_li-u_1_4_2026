# Bài 00: Tạo Series và DataFrame - Bài tập Pokemon
### Giới thiệu:
Trong bài tập này, bạn sẽ tự tay tạo một DataFrame từ dữ liệu dạng Python Dictionary, thay đổi thứ tự các cột, bổ sung cột mới và kiểm tra kiểu dữ liệu của các cột.


### Step 1. Import các thư viện cần thiết

```python
import pandas as pd
import numpy as np
```

### Step 2. Tạo một data dictionary chứa thông tin Pokemon

```python
raw_data = {
    'name': ['Bulbasaur', 'Charmander', 'Squirtle', 'Caterpie'],
    'evolution': ['Ivysaur', 'Charmeleon', 'Wartortle', 'Metapod'],
    'type': ['grass', 'fire', 'water', 'bug'],
    'hp': [45, 39, 44, 45],
    'pokedex': ['yes', 'no', 'yes', 'no']
}
raw_data
```

### Step 3. Gán dictionary vào một biến có tên gọi là pokemon (tạo DataFrame)

```python
pokemon = pd.DataFrame(raw_data)
pokemon
```

### Step 4. Sắp xếp lại thứ tự các cột thành: name, type, hp, evolution, pokedex

```python
pokemon = pokemon[['name', 'type', 'hp', 'evolution', 'pokedex']]
pokemon
```

### Step 5. Thêm một cột mới có tên là place và điền các địa danh tùy ý

```python
pokemon['place'] = ['park', 'mountain', 'lake', 'forest']
pokemon
```

### Step 6. Hiển thị kiểu dữ liệu của từng cột

```python
pokemon.dtypes
```

### BONUS: Câu hỏi tự tạo - Tìm Pokemon có HP cao nhất và lọc các Pokemon hệ cỏ (grass)

```python
# 1. Tìm Pokemon có lượng HP cao nhất
max_hp_pokemon = pokemon[pokemon['hp'] == pokemon['hp'].max()]
print('Pokemon co HP cao nhat:')
print(max_hp_pokemon)

# 2. Loc Pokemon co he la grass
grass_pokemon = pokemon[pokemon['type'] == 'grass']
print('\nPokemon he Grass:')
print(grass_pokemon)
```
