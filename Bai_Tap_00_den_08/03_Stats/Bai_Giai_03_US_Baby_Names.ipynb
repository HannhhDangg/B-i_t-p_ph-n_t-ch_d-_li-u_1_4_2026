# Bài 03: Thống Kê Dữ Liệu (US Baby Names)
### Giới thiệu:
Trong bài tập này, chúng ta sử dụng tập dữ liệu tên trẻ em sinh tại Mỹ (giai đoạn 2004 - 2014) từ Kaggle để thực hành:
- Xóa bỏ các cột thừa (cột index phụ, cột ID vô nghĩa) bằng .drop().
- Đếm phân phối tần suất của biến phân loại (Categorical data) bằng .value_counts().
- Gom nhóm dữ liệu theo tên và tính toán các đại lượng thống kê mô tả: Mean (trung bình), Median (trung vị), Std (độ lệch chuẩn), Min, Max, Quartiles (các phân vị 25%, 50%, 75%).
- Tìm kiếm giá trị cực đại và cực tiểu bằng .idxmax() và mặt nạ Boolean.


### Step 1. Import các thư viện cần thiết

```python
import pandas as pd
import numpy as np
```

### Step 2 & 3. Đọc dữ liệu từ URL và gán vào biến baby_names

```python
url = 'https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/US_Baby_Names_right.csv'
baby_names = pd.read_csv(url)
baby_names.head()
```

### Step 4. Xem 10 dòng đầu tiên của tập dữ liệu

```python
baby_names.head(10)
```

### Step 5. Xóa bỏ hai cột thừa: 'Unnamed: 0' và 'Id'
*Sử dụng drop(columns=...) với axis=1 để loại bỏ cột.*

```python
cols_to_drop = [c for c in ['Unnamed: 0', 'Id'] if c in baby_names.columns]
baby_names.drop(columns=cols_to_drop, inplace=True)
baby_names.head()
```

### Step 6. Trong tập dữ liệu, số lượng tên cho Bé gái (F) hay Bé trai (M) nhiều hơn?

```python
gender_counts = baby_names['Gender'].value_counts()
print(gender_counts)
if gender_counts['F'] > gender_counts['M']:
    print('=> Số lượng tên Bé gái (F) nhiều hơn.')
else:
    print('=> Số lượng tên Bé trai (M) nhiều hơn.')
```

### Step 7. Nhóm tập dữ liệu theo tên (Name) và tính tổng số lượng sinh (Count), gán vào biến names

```python
names = baby_names.groupby('Name')['Count'].sum().reset_index()
names.head()
```

### Step 8. Có bao nhiêu tên khác nhau tồn tại trong tập dữ liệu?

```python
num_unique_names = len(names)
print('Số lượng tên khác nhau trong tập dữ liệu:', num_unique_names)
```

### Step 9. Tên nào xuất hiện nhiều nhất (được đặt nhiều nhất)?

```python
idx_max = names['Count'].idxmax()
most_popular_name = names.loc[idx_max]
print('Tên xuất hiện nhiều nhất:')
print(most_popular_name)
```

### Step 10. Có bao nhiêu tên khác nhau có số lần xuất hiện ít nhất?

```python
min_count = names['Count'].min()
num_least_popular = (names['Count'] == min_count).sum()
print(f'Số lần xuất hiện ít nhất là {min_count}, và có {num_least_popular} tên có số lần xuất hiện này.')
```

### Step 11. Giá trị trung vị (median) của số lần xuất hiện tên là bao nhiêu?

```python
median_count = names['Count'].median()
print('Trung vị (Median) số lần xuất hiện:', median_count)
```

### Step 12. Độ lệch chuẩn (standard deviation) của số lần xuất hiện tên là bao nhiêu?

```python
std_count = names['Count'].std()
print(f'Độ lệch chuẩn (Std): {std_count:.2f}')
```

### Step 13. Lấy bảng tổng kết thống kê mô tả: Mean, Min, Max, Std và các tứ phân vị (Quartiles)

```python
names['Count'].describe()
```
