# Bài 04: Phân Nhóm Dữ Liệu - GroupBy (Tiêu Thụ Đồ Uống Có Cồn)
### Giới thiệu:
Thao tác gom nhóm (groupby) theo mô hình **Split - Apply - Combine** là một trong những công cụ phân tích quan trọng nhất trong Pandas:
- Chia tập dữ liệu lớn thành các nhóm nhỏ theo biến phân loại (ở đây là Châu lục - Continent).
- Áp dụng hàm tổng hợp số học (mean, median, min, max, describe, gg) lên từng nhóm.
- Kết hợp kết quả trả về dưới dạng bảng thống kê so sánh đa chiều.


### Step 1. Import các thư viện cần thiết

```python
import pandas as pd
import numpy as np
```

### Step 2 & 3. Đọc dữ liệu từ URL và gán vào biến drinks

```python
url = 'https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/drinks.csv'
drinks = pd.read_csv(url)
drinks.head()
```

### Step 4. Châu lục nào tiêu thụ nhiều bia (beer_servings) nhất tính theo mức trung bình?

```python
beer_by_continent = drinks.groupby('continent')['beer_servings'].mean()
print(beer_by_continent)
top_beer_continent = beer_by_continent.idxmax()
print(f'\n=> Châu lục uống nhiều bia nhất trung bình: {top_beer_continent} ({beer_by_continent.max():.2f} phần)')
```

### Step 5. Với mỗi châu lục, in ra bảng thống kê mô tả chi tiết mức tiêu thụ rượu vang (wine_servings)

```python
drinks.groupby('continent')['wine_servings'].describe()
```

### Step 6. In ra mức tiêu thụ trung bình (mean) của tất cả các cột số theo từng châu lục
*Lưu ý: Truyền numeric_only=True để tính toán trên các cột số, tránh lỗi khi có cột chuỗi.*

```python
drinks.groupby('continent').mean(numeric_only=True)
```

### Step 7. In ra mức tiêu thụ trung vị (median) của tất cả các cột số theo từng châu lục

```python
drinks.groupby('continent').median(numeric_only=True)
```

### Step 8. In ra các giá trị mean, min và max của mức tiêu thụ rượu mạnh (spirit_servings) theo từng châu lục dưới dạng DataFrame
*Sử dụng phương thức .agg(['mean', 'min', 'max']) để tổng hợp nhiều hàm cùng lúc.*

```python
drinks.groupby('continent')['spirit_servings'].agg(['mean', 'min', 'max'])
```
