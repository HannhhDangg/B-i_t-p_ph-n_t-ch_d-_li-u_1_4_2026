### Giới thiệu:
Trong bài tập này, chúng ta sẽ làm quen với tập dữ liệu đơn hàng của chuỗi nhà hàng Chipotle. Bài tập giúp rèn luyện các kỹ năng đọc dữ liệu từ tệp TSV trên Internet, kiểm tra kích thước, thuộc tính, tính toán các thống kê mô tả cơ bản và làm sạch kiểu dữ liệu chuỗi thành số thực.


### Step 1. Import các thư viện cần thiết

```python
import pandas as pd
import numpy as np
```

### Step 2 & 3. Đọc dữ liệu từ URL và gán vào biến chipo
*Lưu ý: Tệp có định dạng .tsv (Tab-Separated Values) nên cần chỉ định phân cách bằng phím Tab sep='\t'.*

```python
url = 'https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/chipotle.tsv'
chipo = pd.read_csv(url, sep='\t')
chipo.head()
```

### Step 4. Xem 10 dòng đầu tiên của tập dữ liệu

```python
chipo.head(10)
```

### Step 5. Số lượng quan sát (hàng) trong tập dữ liệu là bao nhiêu?

```python
print('Số quan sát (dòng):', chipo.shape[0])
# Cách khác: len(chipo)
```

### Step 6. Số lượng cột trong tập dữ liệu là bao nhiêu?

```python
print('Số cột:', chipo.shape[1])
```

### Step 7. In ra tên của tất cả các cột

```python
print('Danh sách các cột:', list(chipo.columns))
```

### Step 8. Tập dữ liệu được đánh chỉ số (index) như thế nào?

```python
chipo.index
```

### Step 9. Món ăn nào được đặt nhiều nhất?

```python
most_ordered = chipo.groupby('item_name')['quantity'].sum().sort_values(ascending=False)
print('Món được đặt nhiều nhất:', most_ordered.index[0])
```

### Step 10. Với món ăn được đặt nhiều nhất đó, tổng số lượng đã đặt là bao nhiêu?

```python
print('Tổng số lượng của món đó:', most_ordered.iloc[0])
```

### Step 11. Món/phần mô tả kèm theo (choice_description) nào được đặt nhiều nhất?

```python
most_choice = chipo.groupby('choice_description')['quantity'].sum().sort_values(ascending=False)
print('Phần kèm theo được đặt nhiều nhất:', most_choice.index[0])
print('Số lượng:', most_choice.iloc[0])
```

### Step 12. Tổng số lượng tất cả các món đã được đặt là bao nhiêu?

```python
total_items_ordered = chipo['quantity'].sum()
print('Tổng số lượng món đã đặt:', total_items_ordered)
```

### Step 13. Chuyển đổi cột giá tiền (item_price) thành kiểu số thực (float)
#### Step 13.a. Kiểm tra kiểu dữ liệu ban đầu của cột item_price

```python
print('Kiểu dữ liệu ban đầu của item_price:', chipo['item_price'].dtype)
chipo['item_price'].head()
```

#### Step 13.b. Tạo hàm lambda và thay đổi kiểu dữ liệu của item_price

```python
# Loại bỏ ký tự '$' và chuyển đổi sang float
chipo['item_price'] = chipo['item_price'].apply(lambda x: float(x.strip().replace('$', '')))
chipo['item_price'].head()
```

#### Step 13.c. Kiểm tra lại kiểu dữ liệu của item_price sau khi chuyển đổi

```python
print('Kiểu dữ liệu sau chuyển đổi:', chipo['item_price'].dtype)
```

### Step 14. Tổng doanh thu trong toàn bộ thời gian của tập dữ liệu là bao nhiêu?

```python
# Doanh thu = tổng (số lượng * đơn giá)
# Lưu ý: trong dữ liệu Chipotle, item_price chính là tổng giá của dòng đó (item_price = quantity * unit_price).
# Nếu tính tổng doanh thu trực tiếp từ item_price:
revenue = chipo['item_price'].sum()
print(f'Tổng doanh thu: ')
```

### Step 15. Có bao nhiêu đơn hàng (orders) đã được thực hiện?

```python
num_orders = chipo['order_id'].nunique()
print('Số lượng đơn hàng duy nhất:', num_orders)
```

### Step 16. Doanh thu trung bình trên mỗi đơn hàng là bao nhiêu?

```python
avg_revenue_per_order = chipo.groupby('order_id')['item_price'].sum().mean()
print(f'Doanh thu trung bình trên mỗi đơn hàng: ')
```

### Step 17. Có bao nhiêu món ăn khác nhau (khác biệt) được bán?

```python
num_different_items = chipo['item_name'].nunique()
print('Số lượng món ăn khác nhau:', num_different_items)
```
