# Bài 08: Xóa Và Xử Lý Dữ Liệu Thiếu - Deleting & Missing Values (Hoa Iris)
### Giới thiệu:
Tập dữ liệu hoa Iris (Iris Dataset) của nhà sinh vật học Ronald Fisher là một trong những tập dữ liệu kinh điển nhất trong Khoa học Dữ liệu. Bài tập này mô phỏng các tình huống thực tế của quy trình làm sạch dữ liệu (Data Cleaning):
- Đọc dữ liệu thô không có dòng tiêu đề (header=None) và tự định nghĩa tên cột.
- Phát hiện và kiểm tra dữ liệu bị thiếu (isnull() / isna()).
- Cố tình gán giá trị rỗng (
p.nan) vào các vùng dữ liệu cụ thể bằng .iloc.
- Điền khuyết dữ liệu (Imputation) bằng phương thức .fillna().
- Xóa bỏ một cột phân loại bằng del hoặc .drop().
- Loại bỏ các dòng chứa giá trị rỗng bằng .dropna().
- Tái lập chỉ số hàng bằng .reset_index(drop=True).


### Step 1. Import các thư viện cần thiết

```python
import pandas as pd
import numpy as np
```

### Step 2 & 3. Đọc dữ liệu từ URL và gán vào biến iris
*Lưu ý: Tệp iris.data không có dòng tiêu đề cột, do đó cần chỉ định header=None.*

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
iris = pd.read_csv(url, header=None)
iris.head()
```

### Step 4. Đặt tên cột cho tập dữ liệu: ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']

```python
iris.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']
iris.head()
```

### Step 5. Kiểm tra xem có giá trị thiếu (missing values) nào trong DataFrame không?

```python
missing_counts = iris.isnull().sum()
print('Số lượng giá trị thiếu ở mỗi cột:')
print(missing_counts)
```

### Step 6. Gán các giá trị từ dòng 10 đến 29 của cột 'petal_length' thành NaN
*Sử dụng .iloc[10:30, 2] = np.nan (vị trí dòng 10 đến 29 và cột thứ 3 index 2).*

```python
iris.iloc[10:30, 2] = np.nan
print('Kiểm tra giá trị thiếu ở dòng 10 đến 14:')
print(iris.loc[10:14, ['sepal_length', 'petal_length']])
```

### Step 7. Thay thế (điền) các giá trị NaN vừa tạo bằng số 1.0

```python
iris['petal_length'] = iris['petal_length'].fillna(1.0)
print('Kiểm tra lại dòng 10 đến 14 sau khi điền 1.0:')
print(iris.loc[10:14, ['sepal_length', 'petal_length']])
```

### Step 8. Xóa cột 'class' khỏi DataFrame

```python
del iris['class']
# Hoặc: iris.drop(columns=['class'], inplace=True)
iris.head()
```

### Step 9. Gán 3 dòng đầu tiên của DataFrame thành NaN

```python
iris.iloc[0:3, :] = np.nan
iris.head()
```

### Step 10. Xóa tất cả các dòng có chứa giá trị NaN

```python
iris.dropna(axis=0, inplace=True)
iris.head()
```

### Step 11. Đặt lại chỉ số (Reset Index) để chỉ số bắt đầu lại từ 0
*Sử dụng reset_index(drop=True, inplace=True).*

```python
iris.reset_index(drop=True, inplace=True)
iris.head()
```

```python
print('Kích thước DataFrame cuối cùng:', iris.shape)
print('\nThống kê mô tả dữ liệu:')
print(iris.describe())
```
